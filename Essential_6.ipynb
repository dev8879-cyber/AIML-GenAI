#Continuing loops:

```python
for i in range(1,6,2):
  print(i);
```

```python
for i in range(1,6):
  print(i)
```

```python
for i in range(6):
  print(i)
```

```python
for i in range(10,0,-1):
  print(i)
```

```python
for i in range(10,0,-1):
  print(i)
```

```python
#print all even numbers from 1 to 10:
for i in range(0,11,2):
  print(i)
```

```python
#print all odd numbers:
for i in range(1,10,2):
  print(i)
```

```python
for i in "Aditya":
  print(i)
```

```python
for i in "Aditya":
  print(i,end="-")
```

```python
for i in range(2,8):
  print(i*i)
```

```python
for i in range(1,6):
  print("5 *",i,"=",5*i)
```
