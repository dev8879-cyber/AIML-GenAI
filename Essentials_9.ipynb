##Nested Loops:

```python
for i in range(1,6):
  print("i=",i)
```

```python
for i in range(5):
  for j in range(5):
    print("i=",i,"j=",j)
```

```python
for i in range(4):
  for j in range(5):
    print("*",end=" ")
  print()
```

```python
for i in range(1,6):
  for j in range(5):
    print(i,end=" ")
  print()
```

```python
for i in range(5):
  for j in range(1,6):
    print(j,end=" ")
  print()
```

```python
for row in range(1,7):
  for col in range(1,row):
    print("*",end=" ")
  print()
```

```python
for row in range(1,6):
  for col in range(0,row):
    print(row,end=" ");
  print()
```

```python
for i in range(1,6):
  for j in range(1,i+1):
    print(j,end=" ")
  print()
```

```python
n=1;
for i in range(5):
  for j in range(i):
    print(n,end=" ");
    n=n+1
  print()
```

```python
for row in range(4):
  for col in range(4):
    print(col,end=" ")
  print()
```

```python
for i in range(1,3):
  for j in range(1,3):
    print(i+j,end=" ")
  print()
```
